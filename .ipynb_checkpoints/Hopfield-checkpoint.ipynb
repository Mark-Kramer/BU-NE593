Repeat the product of $\xi$ and $S$, **both random**, `K` times.

```{pyodide-python}
K = 100;
square_bracket_term = np.zeros(K)
for k in np.arange(K):
    xi = "SOMETHING (a random pattern)"  # Random pattern
    S  = "SOMETHING (a random pattern)"  # Another random pattern

    square_bracket_term[k] = "SOMETHING"

plt.figure()
plt.hist(square_bracket_term)
plt.axvline(x=N/2, color='red')
plt.axvline(x=0, color='red', linestyle=':')
plt.show()
```

Repeat the product of $\xi$ and $S$, **where $S$ is similar to $\xi$**, `K` times.

```{pyodide-python}
K = 100;
square_bracket_term = np.zeros(K)
for k in np.arange(K):
    xi = "SOMETHING (a random pattern)"  # Random pattern
    S  = "SOMETHING (nearby to xi)"      # A nearby pattern

    square_bracket_term[k] =  "SOMETHING"

plt.figure()
plt.hist(square_bracket_term)
plt.axvline(x=N, color='red')
plt.axvline(x=0, color='red', linestyle=':')
plt.show()

```
